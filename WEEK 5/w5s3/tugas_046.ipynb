Pada percobaan pertama dengan dataset sederhana, bandingkan hasil plot dendrogram single linkage dengan complete linkage dan average linkage.

```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
from scipy.cluster.hierarchy import dendrogram, linkage
from scipy.spatial.distance import pdist, squareform
```

*Penjelasan :*
Kode program melakukan persiapan awal dengan mengimpor perkakas atau library yang akan dipakai dalam seluruh proses analisis data. Pustaka matplotlib.pyplot dipanggil dengan nama alias plt untuk membuat tata letak grafik dan menampilkan visualisasi di layar. Selanjutnya, numpy dipanggil sebagai np untuk mengolah matriks data angka, sedangkan pandas dipanggil sebagai pd untuk pengelolaan data dalam bentuk tabel. Terakhir, fungsi linkage dan dendrogram diambil dari pustaka scipy.cluster.hierarchy untuk menghitung urutan pengelompokan serta menggambar diagram pohonnya, didukung oleh fungsi pdist dan squareform dari scipy.spatial.distance untuk menghitung jarak antar titik data.

```python
# 1. Dataset sederhana
data = np.array(
    [
        [1.0, 1.0],  # a
        [1.5, 1.5],  # b
        [5.0, 5.0],  # c
        [3.0, 4.0],  # d
        [4.0, 4.0],  # e
        [3.0, 3.5],  # f
    ]
)

labels = ["a", "b", "c", "d", "e", "f"]
```

*Penjelasan :*
Kode ini dipakai untuk membuat himpunan data sederhana yang akan dikelompokkan beserta identitas untuk masing-masing titiknya. Variabel data menampung matriks array dua dimensi yang berisi posisi koordinat dari 6 buah titik sampel, yaitu titik $a$ pada titik (1.0, 1.0), titik b pada (1.5, 1.5), titik c pada (5.0, 5.0), titik $d$ pada (3.0, 4.0), titik e pada (4.0, 4.0)$, dan titik $f$ pada (3.0, 3.5). Selain koordinat angka tersebut, dibuat juga variabel labels yang berisi daftar huruf dari "a" hingga "f". Daftar label ini berfungsi untuk memberikan penanda nama pada sumbu bawah grafik dendrogram sehingga kita bisa melihat dengan jelas titik mana saja yang sedang digabungkan.

```python
# 2. Hitung linkage untuk ketiga metode
linked_single = linkage(data, method="single", metric="euclidean")
linked_complete = linkage(data, method="complete", metric="euclidean")
linked_average = linkage(data, method="average", metric="euclidean")
```

*Penjelasan :*
Algoritma klastering hierarki mulai dijalankan untuk menghitung bagaimana titik-titik data tersebut saling bergabung berdasarkan tiga metode pendekatan linkage yang berbeda. Perintah linked_single menghitung pengelompokan berdasarkan jarak terdekat (Single Linkage), linked_complete menghitung pengelompokan berdasarkan jarak terjauh (Complete Linkage), dan linked_average menghitung pengelompokan berdasarkan rata-rata jarak antar anggota (Average Linkage). Ketiga proses perhitungan ini sama-sama menggunakan rumus jarak Euclidean (metric="euclidean") untuk mengukur seberapa dekat atau jauh posisi antar titik data pada bidang cartesian.

```python
# 3. Plot ketiga dendrogram berdampingan
plt.figure(figsize=(18, 5))

# Single Linkage
plt.subplot(1, 3, 1)
dendrogram(linked_single, labels=labels)
plt.title("Single Linkage")
plt.xlabel("Data Points")
plt.ylabel("Distance")
plt.grid(True)

# Complete Linkage
plt.subplot(1, 3, 2)
dendrogram(linked_complete, labels=labels)
plt.title("Complete Linkage")
plt.xlabel("Data Points")
plt.ylabel("Distance")
plt.grid(True)

# Average Linkage
plt.subplot(1, 3, 3)
dendrogram(linked_average, labels=labels)
plt.title("Average Linkage")
plt.xlabel("Data Points")
plt.ylabel("Distance")
plt.grid(True)

plt.tight_layout()
plt.show()
```

*Penjelasan:*
Seluruh hasil kalkulasi dari langkah sebelumnya diubah menjadi tampilan grafik tiga dendrogram yang berdampingan dalam satu area gambar. Pertama, plt.figure(figsize=(18, 5)) membuat kanvas lebar berukuran 18 \times 5 inci, lalu fungsi plt.subplot(1, 3, 1) hingga plt.subplot(1, 3, 3) membagi kanvas menjadi tiga kolom terpisah. Di setiap kolom, fungsi dendrogram() dipanggil untuk menggambar diagram pohon dari masing-masing metode (Single, Complete, dan Average). Setiap grafik dilengkapi dengan judul, label sumbu X (Data Points), label sumbu Y (Distance), serta garis kisi-kisi (plt.grid(True)) sebelum akhirnya dirapikan dengan plt.tight_layout() dan ditampilkan secara utuh menggunakan plt.show().
