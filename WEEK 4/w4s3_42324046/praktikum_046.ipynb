```python
import numpy as np
import pandas as pd
from sklearn.datasets import load_iris
# Load dataset Iris dari scikit-learn
iris = load_iris()
```

*Penjelasan :*
Kode ini melakukan import library yang akan digunakan selama proses klasifikasi. numpy atau np digunakan untuk mengolah data dalam bentuk array atau kumpulan angka. pandas atau pd digunakan untuk mengolah data dalam bentuk tabel atau DataFrame. Sedangkan load_iris digunakan untuk mengambil dataset Iris yang sudah disediakan oleh library Scikit-learn, sehingga tidak perlu membuat atau memasukkan dataset secara manual.

load_iris() digunakan untuk mengambil dataset Iris dari Scikit-learn. Dataset Iris merupakan dataset yang berisi data bunga Iris. Setiap data memiliki empat fitur, yaitu sepal length, sepal width, petal length, dan petal width. Dataset ini juga mempunyai tiga jenis atau spesies bunga, yaitu setosa, versicolor, dan virginica.

```python
# Konversi data menjadi DataFrame Pandas
df = pd.DataFrame(iris.data, columns=iris.feature_names)
df["Species"] = iris.target_names[iris.target]
# Definisikan X (fitur) dan y (target/label)
X = df.drop(columns=["Species"])
y = df["Species"]
print(X)
```

*Penjelasan:*
Data Iris diubah menjadi DataFrame Pandas agar lebih mudah diolah. Kemudian ditambahkan kolom Species yang berisi jenis bunga seperti setosa, versicolor, dan virginica. Setelah itu, data dibagi menjadi X dan y, di mana X berisi empat fitur bunga, sedangkan y berisi jenis/species bunga yang menjadi target prediksi. Terakhir, print(X) digunakan untuk menampilkan data fitur tersebut.

```python
y
```

*Penjelasa:*
Menunjukkan bahwa y berisi label atau nama spesies dari setiap data bunga Iris. Jadi, data nomor 0–4 termasuk setosa, sedangkan data nomor 145–149 termasuk virginica. Totalnya ada 150 data, dan Name: Species menunjukkan bahwa kolom tersebut bernama Species

```python
from sklearn.preprocessing import LabelEncoder
lab = LabelEncoder()
y = lab.fit_transform(y)
y
```

*Penjelasan:*
Pada cell ini, mengimpor dan menggunakan LabelEncoder dari Scikit-Learn untuk melakukan konversi data kategori teks pada variabel y menjadi bentuk angka atau numerik. Proses fit_transform mengubah string nama spesies menjadi angka integer, contohnya setosa menjadi 0, versicolor menjadi 1, dan virginica menjadi 2. Mengubah label menjadi format angka merupakan langkah krusial agar model matematik dapat memproses data label tersebut.

```python
from sklearn.model_selection import train_test_split
X_train, X_test = train_test_split(X, test_size=0.2, random_state=0)
y_train, y_test = train_test_split(y, test_size=0.2, random_state=0)
```

*Penjelasan:*
Fungsi train_test_split dari pustaka Scikit-Learn digunakan untuk membagi dataset menjadi data latih (training) dan data uji (testing). Pembagian ini dilakukan secara terpisah untuk variabel fitur X dan variabel target y. Parameter test_size=0.2 menentukan bahwa 20\% dari total data dialokasikan untuk pengujian (testing), sedangkan 80\% sisanya digunakan untuk pelatihan (training). Parameter random_state=0 dipasang agar proses pengacakan data menghasilkan urutan indeks yang sama setiap kali kode dijalankan, sehingga urutan baris data pada X_train tetap berpasangan secara presisi dengan y_train. 

```python
from sklearn.naive_bayes import GaussianNB
model = GaussianNB()
NB_model = model.fit(X_train, y_train)
```

*Penjelasan:*
Kode ini, dilakukan impor algoritma GaussianNB dari Scikit-Learn. Algoritma Gaussian Naïve Bayes dipilih karena seluruh fitur dalam dataset Iris bersifat kontinu (angka riil/numerik) yang diasumsikan mengikuti distribusi normal (Gaussian). Selanjutnya, untuk melatih model tersebut menggunakan fungsi .fit() dengan memasukkan data latih X_train dan label latih y_train.

```python
result = NB_model.predict(X_test)
result
```

*Penjelasan:*
Kode ini melakukan proses pengujian atau prediksi terhadap data baru yang belum pernah dilihat model sebelumnya, yaitu X_test. Fungsi .predict() akan mengembalikan array berisikan angka kelas (0, 1, atau 2) yang mewakili tebakan spesies model untuk setiap baris data dalam sampel uji X_test.

```python
from sklearn.metrics import accuracy_score
print('Accuracy of data testing: ', accuracy_score(y_test, result)*100)
```

*Penjelasan:*
Kode ini bertujuan untuk mengukur tingkat keberhasilan dan kinerja dari model klasifikasi yang telah dibuat. Dengan membandingkan nilai hasil prediksi (result) terhadap nilai target yang sebenarnya (y_test) menggunakan accuracy_score, diperoleh persentase akurasi model sebesar 96.67\% dalam memprediksi spesies secara tepat.
